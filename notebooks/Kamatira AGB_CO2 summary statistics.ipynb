```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming you've already uploaded and loaded your CSV as df
# df = pd.read_csv('your_file.csv')

# Check the first few rows to understand the data
print(df[['Year', 'Class', 'Sum_AGB', 'Sum_CO2']].head())

# Pivot the data for plotting
pivot_df = df.pivot_table(index=['Year', 'Class'], values=['Sum_AGB', 'Sum_CO2'], aggfunc='sum')

# Reset index for easier plotting
pivot_df = pivot_df.reset_index()

# Plotting
import numpy as np

# Unique years and classes
years = pivot_df['Year'].unique()
classes = pivot_df['Class'].unique()

# Set width of bars
bar_width = 0.35

# Create position for each group
indices = np.arange(len(years))

# Plot for each class
for idx, cls in enumerate(classes):
    # Filter data for this class
    class_data = pivot_df[pivot_df['Class'] == cls]
    # Make sure data is sorted by Year
    class_data = class_data.sort_values('Year')
    # Positions for bars
    position = indices + idx * bar_width
    # Plot bars for Sum_AGB
    plt.bar(position, class_data['Sum_AGB'], width=bar_width, label=f'{cls} - Sum_AGB')
    # Plot bars for Sum_CO2 slightly offset
    plt.bar(position, class_data['Sum_CO2'], width=bar_width, alpha=0.5, label=f'{cls} - Sum_CO2')

# Set labels and titles
plt.xlabel('Year')
plt.ylabel('Sum Values')
plt.title('Sum of AGB and CO2 by Year and Class')
plt.xticks(indices + bar_width / 2 * len(classes), years, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Group by Year and Class, then take the mean
grouped = df.groupby(['Year', 'Class'])[['Mean_AGB', 'Mean_CO2']].mean().reset_index()

# Get unique years and classes
years = sorted(grouped['Year'].unique())
classes = grouped['Class'].unique()

# Prepare data for bar plot: pivot for bar positions
pivot_agb = grouped.pivot(index='Year', columns='Class', values='Mean_AGB')
pivot_co2 = grouped.pivot(index='Year', columns='Class', values='Mean_CO2')

# Number of classes for bar width
n_classes = len(classes)
bar_width = 0.8 / n_classes
x = np.arange(len(years))

plt.figure(figsize=(14, 6))

# Plot bars for each class
for i, cls in enumerate(classes):
    plt.bar(x + i * bar_width, pivot_agb[cls], width=bar_width, label=f'Class {cls} - AGB')

# Plot Mean_CO2 as a line
# To plot the line, take the mean of Mean_CO2 across classes for each year
mean_co2_per_year = grouped.groupby('Year')['Mean_CO2'].mean()
plt.plot(x + bar_width * (n_classes - 1) / 2, mean_co2_per_year, color='red', marker='o', label='Mean CO2')

# Labels and title
plt.xlabel('Year')
plt.ylabel('Values')
plt.title('Mean AGB per Class (Grouped Bars) and Mean CO2 (Line) over Years')
plt.xticks(x + bar_width * (n_classes - 1) / 2, years, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Updated class names
class_colors = {
    'Dense Forest Class': '#004d00',  # dense use
    'Moderate Forest Class': '#33cc33',  # moderate use
    'Open Forest Class': '#66ff66'  # open use
}

# Map class to colors
df['Color'] = df['Class'].map(class_colors)

# Unique classes for plotting
classes = ['Dense Forest Class', 'Moderate Forest Class', 'Open Forest Class']

# Get unique years for x-axis, sorted
years = sorted(df['Year'].unique())

# Create a figure
plt.figure(figsize=(10,6))

# Set width for bars and offset for clustering
bar_width = 1.0
offsets = {
    'Dense Forest Class': -bar_width,
    'Moderate Forest Class': 0,
    'Open Forest Class': bar_width
}

# Plot bars for Mean_AGB
for cls in classes:
    subset = df[df['Class'] == cls]
    plt.bar(
        [year + offsets[cls] for year in subset['Year']],
        subset['Mean_AGB'],
        width=bar_width,
        color=class_colors[cls],
        label=f'{cls} AGB'
    )

# Plot lines for Mean_CO2
for cls in classes:
    subset = df[df['Class'] == cls]
    plt.plot(
        subset['Year'] + offsets[cls],
        subset['Mean_CO2'],
        color='blue' if cls=='Dense Forest Class' else 'orange' if cls=='Moderate Forest Class' else 'red',
        marker='o',
        label=f'{cls} CO2'
    )

# Add labels and title
plt.xlabel('Year')
plt.ylabel('AGB and CO2 by Forest Class in Tons')
plt.title('Mean AGB (bars) and Mean CO2 (lines) by Year and Class in Tons')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Select relevant columns with exact names
df = df_agb[['Year', 'Min   (Mg/ha)', 'Max  (Mg/ha)', 'Mean  (Mg/ha)', 'StdDev']]

# Define metrics and colors
metrics = ['Min   (Mg/ha)', 'Max  (Mg/ha)', 'Mean  (Mg/ha)', 'StdDev']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

# Set bar width
bar_width = 0.2
x = np.arange(len(df['Year']))

plt.figure(figsize=(15, 6))

# Plot each metric as a clustered bar
for i, (metric, color) in enumerate(zip(metrics, colors)):
    bars = plt.bar(x + i * bar_width, df[metric], width=bar_width, label=metric, color=color)
    # Overlay the value labels on each bar
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height,
                 f'{height:.1f}', ha='center', va='bottom', fontsize=8)

# Set labels and title
plt.xlabel('Year')
plt.ylabel('AGB (Mg/ha')
plt.title('AGB Summary Statistics Over Years in Mg/ha')
plt.xticks(x + bar_width * (len(metrics) - 1) / 2, df['Year'].astype(int).astype(str), rotation=45)

# Set x-axis limits
plt.xlim(-0.5, len(df['Year']) - 0.5)

# Add legend
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```
